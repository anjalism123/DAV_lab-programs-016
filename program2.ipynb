```python
import pandas as pd

df = pd.read_csv('students.csv')

print(df.describe())

print('\nMissing values:\n', df.isnull().sum())

print('\nDuplicate rows:', df.duplicated().sum())

Fee = pd.DataFrame({
    'Branch': ['CSE', 'ECE', 'ISE'],
    'Annual Fee': [120000, 110000, 115000]
})

merged = pd.merge(df, Fee, on='Branch', how='left')

print(merged)

print('\nAverage Marks by Branch:\n', df.groupby('Branch')['Marks'].mean())

pivot = pd.pivot_table(
    df,
    values='Marks',
    index='Branch',
    columns='Result',
    aggfunc='count'
)

print(pivot)
```
