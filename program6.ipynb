```python

import pandas as pd
import numpy as np

from sklearn.preprocessing import (
    MinMaxScaler,
    StandardScaler,
    LabelEncoder
)

from sklearn.impute import SimpleImputer

data = {
    'Age': [20, 25, 30, 35, 40, np.nan],
    'Salary': [20000, 30000, 40000, 50000, 60000, 70000],
    'Experience': [1, 3, 5, 7, 9, 11],
    'Department': [
        'IT', 'HR', 'IT', 'Finance', 'HR', 'Finance'
    ],
    'Purchased': [0, 1, 0, 1, 1, 0]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

imputer = SimpleImputer(strategy='mean')

df[['Age', 'Salary', 'Experience']] = (
    imputer.fit_transform(
        df[['Age', 'Salary', 'Experience']]
    )
)

# Create a new feature: Annual Salary per Experience Year

df['Salary_Per_Experience'] = (
    df['Salary'] / (df['Experience'] + 1)
)

# Create Age Group Feature
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[0, 25, 35, 100],
    labels=['Young', 'Middle', 'Senior'],
    include_lowest=True
)

print("\nAfter Feature Engineering:")
print(df)

# Convert categorical columns into numerical columns

df = pd.get_dummies(
    df,
    columns=['Department', 'Age_Group'],
    dtype=int
)

print("\nAfter Feature Encoding:")
print(df)

X = df.drop('Purchased', axis=1)
y = df['Purchased']

# Scale values between 0 and 1

normalizer = MinMaxScaler()

X_normalized = normalizer.fit_transform(X)

normalized_df = pd.DataFrame(
    X_normalized,
    columns=X.columns
)

print("\nNormalized Dataset (0 to 1):")
print(normalized_df.round(3))

# Mean = 0 and Standard Deviation = 1

standardizer = StandardScaler()

X_standardized = standardizer.fit_transform(X)

standardized_df = pd.DataFrame(
    X_standardized,
    columns=X.columns
)

print("\nStandardized Dataset:")
print(standardized_df.round(3))

print("\n========== FINAL COMPARISON ==========")

print("\nOriginal Features:")
print(X.head())

print("\nNormalized Features:")
print(normalized_df.head())

print("\nStandardized Features:")
print(standardized_df.head())

print("\nTarget Variable:")
print(y)
```
