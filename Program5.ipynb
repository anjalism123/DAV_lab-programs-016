```python
import pandas as pd

df = pd.read_csv('students.csv')

q1 = df['StudyHours'].quantile(0.25)
q3 = df['StudyHours'].quantile(0.75)

IQR = q3 - q1
lower = q1 - 1.5 * IQR
upper = q3 + 1.5 * IQR

outlier_iqr = df[
    (df['StudyHours'] < lower) |
    (df['StudyHours'] > upper)
]

print(f'IQR bounds: [{lower:.2f}, {upper:.2f}]')
print('Outliers found (IQE):', len(outlier_iqr))

#### Z-Score ####

df['Z_Score'] = (
    (df['StudyHours'] - df['StudyHours'].mean())
    / df['StudyHours'].std()
)

outliers_z = df[df['Z_Score'].abs() > 3]

print('Outliers found (Z-Score>3):', len(outliers_z))

#### Capping ####

df['StudyHours_Capped'] = df['StudyHours'].clip(
    lower=lower,
    upper=upper
)

print('\nBefore vs After capping(max value):')
print('Before:', df['StudyHours'].max())
print('After :', df['StudyHours_Capped'].max())
```
